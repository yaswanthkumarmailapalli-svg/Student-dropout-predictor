```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)

print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)

print("Testing target:", y_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
categorical_transformer = OneHotEncoder(
    handle_unknown='ignore'
)
```

```python
numerical_transformer = "passthrough"
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('categorical', categorical_transformer, categorical_features),
        ('numerical', numerical_transformer, numerical_features)
    ]
)
```

```python
print("Preprocessor created successfully!")

print(preprocessor)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])

print("Random Forest model created successfully!")
```

```python
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
X_train_transformed = rf_model.named_steps["preprocessor"].transform(X_train)

print("Original training data shape:", X_train.shape)
print("Transformed training data shape:", X_train_transformed.shape)
```

```python
y_pred = rf_model.predict(X_test)

print("First 10 actual outcomes:")
print(y_test.head(10).values)

print("\nFirst 10 predicted outcomes:")
print(y_pred[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

import matplotlib.pyplot as plt
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Random Forest Accuracy:", round(accuracy * 100, 2), "%")
```

```python
print("Random Forest Classification Report:\n")

print(classification_report(
    y_test,
    y_pred,
    labels=["Dropout", "Enrolled", "Graduate"],
    zero_division=0
))
```

```python
labels = ["Dropout", "Enrolled", "Graduate"]

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(cmap="Blues")

plt.title("Random Forest Confusion Matrix")
plt.show()
```

```python
rf_balanced = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

print("Balanced Random Forest model created!")
```

```python
rf_balanced.fit(X_train, y_train)

print("Balanced Random Forest trained successfully!")
```

```python
y_pred_balanced = rf_balanced.predict(X_test)

print("First 10 actual outcomes:")
print(y_test.head(10).values)

print("\nFirst 10 balanced model predictions:")
print(y_pred_balanced[:10])
```

```python
print("Balanced Random Forest Accuracy:",
      round(accuracy_score(y_test, y_pred_balanced) * 100, 2), "%")

print("\nClassification Report:\n")

print(classification_report(
    y_test,
    y_pred_balanced,
    labels=["Dropout", "Enrolled", "Graduate"],
    zero_division=0
))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

comparison = pd.DataFrame({
    "Model": ["Random Forest", "Balanced Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_balanced)
    ],
    "Dropout Precision": [
        precision_score(y_test, y_pred, labels=["Dropout"], average="macro", zero_division=0),
        precision_score(y_test, y_pred_balanced, labels=["Dropout"], average="macro", zero_division=0)
    ],
    "Dropout Recall": [
        recall_score(y_test, y_pred, labels=["Dropout"], average="macro", zero_division=0),
        recall_score(y_test, y_pred_balanced, labels=["Dropout"], average="macro", zero_division=0)
    ],
    "Dropout F1-score": [
        f1_score(y_test, y_pred, labels=["Dropout"], average="macro", zero_division=0),
        f1_score(y_test, y_pred_balanced, labels=["Dropout"], average="macro", zero_division=0)
    ]
})

print(comparison.round(4))
```

```python
from sklearn.inspection import permutation_importance

importance = permutation_importance(
    rf_balanced,
    X_test,
    y_test,
    n_repeats=5,
    random_state=42,
    scoring="f1_macro",
    n_jobs=-1
)

feature_importance = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": importance.importances_mean
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("Top 15 Important Features:\n")
print(feature_importance.head(15))
```

```python
top_features = feature_importance.head(15)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title("Top 15 Important Features - Random Forest")
plt.xlabel("Permutation Importance")
plt.ylabel("Features")
plt.show()
```

```python
y_prob = rf_balanced.predict_proba(X_test)

print("First 10 probability predictions:\n")
print(y_prob[:10])

print("\nClass order:")
print(rf_balanced.classes_)
```

```python
dropout_index = list(rf_balanced.classes_).index("Dropout")

dropout_probability = y_prob[:, dropout_index]

print("First 10 dropout probabilities:\n")
print(dropout_probability[:10])
```

```python
risk_df = pd.DataFrame({
    "Student_Index": X_test.index,
    "Dropout_Probability": dropout_probability,
    "Predicted_Outcome": y_pred_balanced
})

risk_df = risk_df.sort_values(
    by="Dropout_Probability",
    ascending=False
)

risk_df["Rank"] = range(1, len(risk_df) + 1)

risk_df = risk_df[
    ["Rank", "Student_Index", "Dropout_Probability", "Predicted_Outcome"]
]

print("Top 20 students with highest predicted dropout probability:\n")
print(risk_df.head(20))
```

```python
risk_df.to_csv("Random_Forest_Dropout_Risk_Ranking.csv", index=False)

print("Student risk ranking saved successfully!")
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
```

```python
rf_tuning_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        random_state=42,
        n_jobs=1
    ))
])

print("Tuning pipeline created successfully!")
```

```python
param_grid = {
    "classifier__n_estimators": [200, 300, 500],
    "classifier__max_depth": [None, 10, 20, 30, 40],
    "classifier__min_samples_split": [2, 5, 10],
    "classifier__min_samples_leaf": [1, 2, 4, 6],
    "classifier__max_features": ["sqrt", "log2", 0.7],
    "classifier__class_weight": [None, "balanced", "balanced_subsample"]
}

print("Parameter combinations defined!")
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
rf_search = RandomizedSearchCV(
    estimator=rf_tuning_pipeline,
    param_distributions=param_grid,
    n_iter=20,
    scoring="f1_macro",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_search.fit(X_train, y_train)

print("Hyperparameter tuning completed!")
```

```python
print("Best cross-validation F1 Macro Score:")
print(round(rf_search.best_score_, 4))

print("\nBest parameters:")
print(rf_search.best_params_)
```

```python
best_rf_model = rf_search.best_estimator_

print("Best Random Forest model selected!")
```

```python
y_pred_tuned = best_rf_model.predict(X_test)

print("First 10 actual outcomes:")
print(y_test.head(10).values)

print("\nFirst 10 tuned predictions:")
print(y_pred_tuned[:10])
```

```python
print("Tuned Random Forest Accuracy:",
      round(accuracy_score(y_test, y_pred_tuned) * 100, 2), "%")

print("\nTuned Random Forest Classification Report:\n")

print(classification_report(
    y_test,
    y_pred_tuned,
    labels=["Dropout", "Enrolled", "Graduate"],
    zero_division=0
))
```

```python
print("Tuned Random Forest Accuracy:",
      round(accuracy_score(y_test, y_pred_tuned) * 100, 2), "%")

print("\nTuned Random Forest Classification Report:\n")

print(classification_report(
    y_test,
    y_pred_tuned,
    labels=["Dropout", "Enrolled", "Graduate"],
    zero_division=0
))
```

```python
labels = ["Dropout", "Enrolled", "Graduate"]

cm_tuned = confusion_matrix(
    y_test,
    y_pred_tuned,
    labels=labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_tuned,
    display_labels=labels
)

disp.plot(cmap="Blues")

plt.title("Tuned Random Forest Confusion Matrix")
plt.show()
```

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline
```

```python
rf_oversample = ImbPipeline(steps=[
    ("preprocessor", preprocessor),
    ("oversampler", RandomOverSampler(random_state=42)),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ))
])

print("Oversampling pipeline created!")
```

```python
rf_oversample.fit(X_train, y_train)

print("Oversampled Random Forest trained successfully!")
```

```python
y_pred_oversample = rf_oversample.predict(X_test)

print("Oversampled Random Forest predictions completed!")
```

```python
print("Oversampled Random Forest Accuracy:",
      round(accuracy_score(y_test, y_pred_oversample) * 100, 2), "%")

print("\nClassification Report:\n")

print(classification_report(
    y_test,
    y_pred_oversample,
    labels=["Dropout", "Enrolled", "Graduate"],
    zero_division=0
))
```

```python
models = {
    "Original Random Forest": y_pred,
    "Balanced Random Forest": y_pred_balanced,
    "Tuned Random Forest": y_pred_tuned,
    "Oversampled Random Forest": y_pred_oversample
}

comparison_results = []

for model_name, predictions in models.items():

    report = classification_report(
        y_test,
        predictions,
        output_dict=True,
        zero_division=0
    )

    comparison_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Dropout Precision": report["Dropout"]["precision"],
        "Dropout Recall": report["Dropout"]["recall"],
        "Dropout F1": report["Dropout"]["f1-score"],
        "Enrolled Precision": report["Enrolled"]["precision"],
        "Enrolled Recall": report["Enrolled"]["recall"],
        "Enrolled F1": report["Enrolled"]["f1-score"],
        "Graduate Precision": report["Graduate"]["precision"],
        "Graduate Recall": report["Graduate"]["recall"],
        "Graduate F1": report["Graduate"]["f1-score"],
        "Macro F1": report["macro avg"]["f1-score"]
    })

comparison_df = pd.DataFrame(comparison_results)

print(comparison_df.round(3).to_string(index=False))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Macro F1"
)

plt.title("Comparison of Random Forest Models")
plt.xlabel("Model")
plt.ylabel("Macro F1-score")
plt.xticks(rotation=20)
plt.show()
```
